```python
def main():
    car_rented = [False, False, False]  # [0]: Tesla, [1]: Hyundai, [2]: Ford

    print("\n\n\n\n\n\n\n\n\t\t\t       CAR RENTAL SYSTEM \n\n")
    customername = input("Enter the customer name : ")

    # Simulating the C++ 'goto car;' loop structure
    while True:
        print("--------------------------------------------------------------------------")
        print("Press A for Tesla \nPress B for Hyundai \nPress C for Ford :")
        carmodel = input()
        print("\n\n--------------------------------------------------------------------------")

        if carmodel == "A" or carmodel == "a":
            if car_rented[0]:
                print("Tesla is already rented. Choose another car.")
                continue
            print("You have chosen Tesla ")
            model = int(input("Enter model you want:"))
            if 2000 < model < 2010:
                print("its an old tesla that costs the following per day: 700", end="")
            elif 2010 <= model <= 2023:
                print("its a new tesla that costs the following per days:1500", end="")
            car_rented[0] = True  # Mark Tesla as rented
            break

        elif carmodel == "B" or carmodel == "b":
            if car_rented[1]:
                print("Hyundai is already rented. Choose another car.")
                continue
            print("You have chosen Hyundai ")
            model = int(input("Enter model you want:"))
            if 2000 < model < 2010:
                print("its an old hyundai that costs the following per day:800", end="")
            elif 2010 <= model <= 2023:
                print("its a new huandai that costs the following per day:2000", end="")
            car_rented[1] = True  # Mark Hyundai as rented
            break

        elif carmodel == "C" or carmodel == "c":
            if car_rented[2]:
                print("Ford is already rented. Choose another car.")
                continue
            print("You have chosen Ford ")
            model = int(input("Enter model you want:"))
            if 2000 < model < 2010:
                print("its an old ford that costs the following per day:1000", end="")
            elif 2010 <= model <= 2023:
                print("its a new ford that costs the following per day:2200", end="")
            car_rented[2] = True  # Mark Ford as rented
            break

        else:
            print("You have not selected from our library")
            continue

    print("\n\n--------------------------------------------------------------------------")

    print("\n\tCAR Number You have \n\nADS:328__AXS:385_ADJ:146\n")
    carnumber = input("Enter the car number  : \n")

    days = int(input("Enter the Days you want : "))

    # Exact logic replication from C++ code (including operator precedence)
    rentalfee = 0
    if (carmodel == "A" or carmodel == "a") and (2000 < model < 2010):
        rentalfee = days * 700
    elif (carmodel == "A" or carmodel == "a") and (2010 <= model <= 2023):
        rentalfee = days * 1400

    if (carmodel == "B" or carmodel == "b") and (2000 < model < 2010):
        rentalfee = days * 800
    elif (carmodel == "B" or carmodel == "b") and (2010 <= model <= 2023):
        rentalfee = days * 2000

    if (carmodel == "C" or carmodel == "c") and (2000 < model < 2010):
        rentalfee = days * 1000
    elif (carmodel == "C" or carmodel == "c") and (2010 <= model <= 2023):
        rentalfee = days * 2200

    print("\n\t\t                       Car Rental - Customer Invoice                  ")
    print("\t\t\t///////////////////////////////////////////////////////////")
    print(f"\t\t\t| Invoice No. :------------------|{'#Cnb81353':>10} |")
    print(f"\t\t\t| Customer Name:-----------------|{customername:>10} |")
    print(f"\t\t\t| Car Model :--------------------|{carmodel:>10} |")
    print(f"\t\t\t| Car No. :----------------------|{carnumber:>10} |")
    print(f"\t\t\t| Number of days :---------------|{days:>10} |")
    print(f"\t\t\t| Your Rental Amount is :--------|{rentalfee:>10} |")
    print("\t\t\t __")
    print()
    print(f"\t\t\t| Total Rental Amount is :-------|{rentalfee:>10} |")
    print("\t\t\t __")
    print("\t\t\t # This is a computer generated invoice and it does not")
    print("\t\t\t require an authorised signature #")
    print(" ")
    print("\t\t\t///////////////////////////////////////////////////////////")
    print("\t\t\tYou are advised to pay up the amount before due date.")
    print("\t\t\tOtherwise penelty fee will be applied")
    print("\t\t\t///////////////////////////////////////////////////////////")

    delete_option = input("Do you want to delete data of rented cars? (Y/N): ")

    if delete_option == 'Y' or delete_option == 'y':
        if carmodel == "A" or carmodel == "a":
            car_rented[0] = False
        elif carmodel == "B" or carmodel == "b":
            car_rented[1] = False
        elif carmodel == "C" or carmodel == "c":
            car_rented[2] = False

        print("Data of rented cars deleted successfully.")

if __name__ == "__main__":
    main()
```
